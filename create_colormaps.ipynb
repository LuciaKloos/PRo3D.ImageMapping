```python
from colorspacious import cspace_converter

import matplotlib.pyplot as plt
import numpy as np

import matplotlib as mpl

```

```python
cmaps = {}

gradient = np.linspace(0, 1, 256)
gradient = np.vstack((gradient, gradient))


def plot_color_gradients(category, cmap_list, save_path=None):
    nrows = len(cmap_list)
    figh = 0.35 + 0.15 + (nrows + (nrows - 1) * 0.1) * 0.22
    fig, axs = plt.subplots(nrows=nrows, figsize=(6.4, figh))
    if nrows == 1:
        axs = [axs]  # Sicherstellen, dass axs iterierbar ist

    fig.subplots_adjust(left=0, right=1, top=1, bottom=0, hspace=0)

    for ax, name in zip(axs, cmap_list):
        ax.imshow(gradient, aspect='auto', cmap=mpl.colormaps[name])
        ax.set_axis_off()  # alles ausschalten

    # Speichern ohne Weißraum
    if save_path:
        fig.savefig(save_path, bbox_inches='tight', pad_inches=0)
    
    plt.close(fig)  # Schließt die Figur, um Speicher zu sparen

    # Colormap-Liste speichern
    cmaps[category] = cmap_list

# Beispiel-Aufruf
```

```python
# plot_color_gradients('Cyclic', ['twilight_shifted'])
plot_color_gradients("Perceptually Uniform Sequential", ["plasma"], save_path="plasma.png")
```
